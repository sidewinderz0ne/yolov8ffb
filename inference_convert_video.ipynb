```python
# Importing required modules
import subprocess
import os
from tqdm import tqdm  # Progress bar library

# Directly specify the folder path containing video files
folder_path = r"C:\Users\DELL\Documents\video\test"  # Replace with your actual folder path

# Check if the folder exists
if not os.path.isdir(folder_path):
    raise FileNotFoundError(f"The specified folder {folder_path} does not exist.")

# Get the list of video files from the folder
video_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith(('.mp4', '.avi'))]

# List of parameter arrays
# model, image size, roi, conf, iou
parameter_list = [
    [r"C:\Users\DELL\Documents\model\yolov5nu_1280_sampling_only_revisi\best.pt", "800", "0.1", "0.43", "0.3"],
    [r"C:\Users\DELL\Documents\model\yolov5nu_1280_sampling_only_revisi\best.pt", "800", "0.1", "0.43", "0.3"],
    [r"C:\Users\DELL\Documents\model\yolov5nu_1280_sampling_only_revisi\best.pt", "1280", "0.2", "0.43", "0.3"],
    [r"C:\Users\DELL\Documents\model\yolov5nu_1280_sampling_only_revisi\best.pt", "1024", "0.2", "0.43", "0.3"],
]

# Check if all model paths are valid
for params in parameter_list:
    if not os.path.isfile(params[0]):
        raise FileNotFoundError(f"The specified model file {params[0]} does not exist or is not accessible.")

# Total number of iterations
total_iterations = len(video_files) * len(parameter_list)

# Using tqdm for progress tracking
with tqdm(total=total_iterations, desc="Processing", unit="task") as pbar:
    # Iterating through each parameter array and running the script for each video file
    for video_file in video_files:
        for params in parameter_list:
            command = [
                "python", "11-track-master.py",
                "--yolo_model", params[0],
                "--imgsz", params[1],
                "--roi", params[2],
                "--conf_thres", params[3],
                "--iou_thres", params[4],
                "--source", video_file,  # Using video file from the folder
                "--mode", "testing",  # Replace or modify as needed
                "--save_vid", "True"  # Replace or modify as needed
            ]

            # Run the command
            try:
                subprocess.run(command, check=True)
                print(f"\nCommand run successfully for model {params[0]} on video {video_file}")
            except subprocess.CalledProcessError as e:
                print(f"\nError occurred while running command for model {params[0]} on video {video_file}: {e}")
            
            # Update progress bar
            pbar.update(1)
```
